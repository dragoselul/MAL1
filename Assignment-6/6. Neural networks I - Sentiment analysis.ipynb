# Sentiment Analysis Implementation Documentation

## Overview
This implementation performs sentiment analysis on movie reviews using a neural network. The system classifies text as either positive or negative sentiment.

## Prerequisites
- Python 3.x
- Required libraries: numpy, pandas, scikit-learn
- Data files: 'reviews.txt' (containing review text), 'labels.txt' (containing sentiment labels)

## Code Structure

### 1. Data Preparation
```python
reviews = pd.read_csv('reviews.txt', header=None)
labels = pd.read_csv('labels.txt', header=None)
Y = (labels == 'positive').astype(np.int_)
```
- Loads review text and labels from files
- Converts labels to binary format (0 for negative, 1 for positive)

### 2. Data Splitting
```python
X_temp, X_test, y_temp, y_test = train_test_split(
    reviews[0], Y[0], test_size=0.2, random_state=42
)
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=42
)
```
- Creates three datasets:
  - Training (60%): Used to train the model
  - Validation (20%): Used for hyperparameter tuning
  - Test (20%): Used for final evaluation
- `random_state=42` ensures reproducibility

### 3. Text Vectorization
```python
vectorizer = CountVectorizer(max_features=10000)
X_train_bow = vectorizer.fit_transform(X_train)
```
- Converts text to numerical format using Bag-of-Words
- `max_features=10000`: Keeps only the 10,000 most frequent words
- Important: Only fit the vectorizer on training data to prevent data leakage

### 4. Neural Network Configuration
```python
model = MLPClassifier(
    hidden_layer_sizes=(100,),
    activation='relu',
    solver='adam',
    alpha=0.0001,
    batch_size='auto',
    learning_rate='adaptive',
    max_iter=200
)
```
Parameters explained:
- `hidden_layer_sizes=(100,)`: One hidden layer with 100 neurons
- `activation='relu'`: ReLU activation function prevents vanishing gradients
- `solver='adam'`: Efficient optimizer for large datasets
- `alpha=0.0001`: L2 regularization strength
- `learning_rate='adaptive'`: Adjusts learning rate during training
- `max_iter=200`: Maximum training iterations

## Best Practices

   - Always split data before vectorization

   - Use consistent random seeds

2. Model Training
   - Monitor validation accuracy for overfitting

   - Gradually increase complexity if needed

3. Evaluation
   - Use multiple metrics (not just accuracy)
   - Consider class balance
   - Test with diverse examples

## Common Issues and Solutions

1. Poor Performance
   - Check data preprocessing
   - Increase model complexity
   - Adjust regularization strength

2. Overfitting
   - Reduce network size
   - Increase regularization
   - Add dropout layers

3. Slow Training
   - Reduce vocabulary size
   - Decrease batch size
   - Adjust learning rate

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
````

````python
# Already performed, but helpful to see the process
# reviews = pd.read_csv('reviews.txt', header=None)
# labels = pd.read_csv('labels.txt', header=None)
# Y = (labels == 'positive').astype(np.int_)
````

````python
# Split into train+validation and test sets
X_temp, X_test, y_temp, y_test = train_test_split(
    reviews[0], Y[0], test_size=0.2, random_state=42
)

# Split remaining data into train and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=42
)

# Create Bag-of-Words representation
vectorizer = CountVectorizer(max_features=10000)
X_train_bow = vectorizer.fit_transform(X_train)
X_val_bow = vectorizer.transform(X_val)
X_test_bow = vectorizer.transform(X_test)

print(f"Training set size: {X_train_bow.shape}")
print(f"Validation set size: {X_val_bow.shape}")
print(f"Test set size: {X_test_bow.shape}")
````

````python
# Get vocabulary
vocab = vectorizer.get_feature_names_out()
print("Vocabulary size:", len(vocab))
print("\nFirst 10 words in vocabulary:", vocab[:10])

# Analyze a single review
sample_review = X_train_bow[0]
print("\nShape of a single review vector:", sample_review.shape)
nonzero_elements = sample_review.nonzero()[1]
words_present = [(vocab[i], sample_review[0, i]) for i in nonzero_elements]
print("\nWords in first review (word, count):")
for word, count in words_present[:10]:
    print(f"{word}: {count}")
````

````python
model = MLPClassifier(
    hidden_layer_sizes=(100,),
    activation="logistic",
    solver='adam',
    alpha=0.0001,
    batch_size='auto',
    learning_rate='adaptive',
    max_iter=200,
    random_state=42
)

model.fit(X_train_bow, y_train)
val_pred = model.predict(X_val_bow)
val_accuracy = accuracy_score(y_val, val_pred)
print("\nValidation Accuracy:", val_accuracy)
````

````python
test_pred = model.predict(X_test_bow)
test_accuracy = accuracy_score(y_test, test_pred)
print("\nTest Set Performance:")
print("Accuracy:", test_accuracy)
print("\nDetailed Classification Report:")
print(classification_report(y_test, test_pred))
````

````python
def classify_review(review_text):
    # Transform the review
    review_bow = vectorizer.transform([review_text])
    # Get prediction and probability
    prediction = model.predict(review_bow)[0]
    probability = model.predict_proba(review_bow)[0]
    
    sentiment = "positive" if prediction == 1 else "negative"
    confidence = probability[1] if prediction == 1 else probability[0]
    
    return sentiment, confidence

# Example usage
example_reviews = [
    "This movie was absolutely fantastic!",
    "I really hated every minute of this film.",
    "The plot was okay, but the acting could have been better."
    "An absolute masterpiece, I loved every second of it.",
    "The worst movie I have ever seen, do not waste your time.",
    "It was an average film, nothing special but not terrible either.",
    "The storyline was captivating and the characters were well-developed.",
    "I couldn't stand the acting, it was so bad.",
    "A beautiful film with a touching story.",
    "Terrible plot and poor execution, I wouldn't recommend it.",
    "A delightful movie that kept me entertained from start to finish.",
    "The special effects were amazing, but the story was lacking.",
    "A boring and predictable film, I fell asleep halfway through.",
    "An inspiring movie with a powerful message.",
    "The dialogue was cringe-worthy and the acting was subpar.",
    "A visually stunning film with a weak plot.",
    "I enjoyed the movie, but it had some flaws.",
    "A fantastic film that I would watch again.",
    "The movie was okay, but it didn't live up to the hype.",
    "A complete waste of time, I regret watching it.",
    # Intentionally vague reviews using mismatched sentiment
    "Its a good thing I didn't pay to see this movie.",
    "I wish I could get back the time I spent watching this film.",
    "I loved every minute of this terrible movie.",
    "I hated this film, it was so good!",
    "It was a terrible movie, I loved it!",
    "This movie feels like the masterpiece of an elementary school dropout.",
    "This movie was a change from the awful sequels we've been getting.",
]

print("\nTesting custom reviews:")
for review in example_reviews:
    sentiment, confidence = classify_review(review)
    print(f"\nReview: {review}")
    print(f"Sentiment: {sentiment}")
    print(f"Confidence: {confidence:.2f}")
````
